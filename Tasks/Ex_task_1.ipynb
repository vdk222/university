**Пример:**


**Пункт 12:**

Используя `filter()` и `lambda`, из списка чисел оставьте только простые числа.

**Пример:**

```python
# Входные данные
numbers = [2, 3, 4, 5, 6, 7, 8, 9, 10]

# Ожидаемый результат
[2, 3, 5, 7]
```

*Подсказка:* Для проверки числа на простоту может понадобиться дополнительная функция.


````python
numbers = [2, 3, 4, 5, 6, 7, 8, 9, 10]
def prst(n):
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

numbers = [2, 3, 4, 5, 6, 7, 8, 9, 10]

result = list(filter(lambda x: prst(x), numbers))
print(result)
````

**Пункт 14:**

С помощью `reduce()` и `lambda` подсчитайте количество появлений каждого символа в строке (без учёта регистра), и выведите символ, который встречается чаще всего.

**Пример:**

```python
# Входные данные
text = 'Functional programming with lambda functions'

# Ожидаемый результат
'n'  # n встречается чаще всего
```

*Подсказка:* Возможно, понадобится преобразовать строку в список символов и использовать словарь для подсчёта.

````python
from functools import reduce

text = 'Functional programming with lambda functions'
f = reduce(lambda x, y: {**x, y: x.get(y, 0) + 1}, text.lower(),{})
res = max(f, key=f.get)
print(res)
````

**Пункт 15:**

Используя `map()`, `filter()`, `reduce()` и `lambda`, из заданного списка слов найдите общее количество гласных букв.

**Пример:**

```python
# Входные данные
words = ['python', 'lambda', 'functions', 'map', 'filter', 'reduce']

# Ожидаемый результат
12
```

*Подсказка:* Гласные буквы — 'a', 'e', 'i', 'o', 'u'.

````python
from functools import reduce

words = ['python', 'lambda', 'functions', 'map', 'filter', 'reduce']
result = reduce(lambda x, y: x + y, map(lambda word: len(list(filter(lambda c: c in 'aeiou', word))), words))
print(result)
````
